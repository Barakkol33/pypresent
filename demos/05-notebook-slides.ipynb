```python
from pypresent import Presentation, code, figure, result, slide

deck = Presentation(
    slides='05-notebook-slides.ipynb',
    source='05-notebook-lecture.ipynb',
    output='out/05-notebook.html',
    title='A deck that quotes its lecture',
    date='A pypresent demo',
    theme='office',
)
deck
```

```python
slide('''
# A deck that quotes its lecture

Nothing on these slides was pasted.
''')
```

```python
slide('''
## The cell, quoted by name

- the lecture names it `# slide: tokenize`
- `trim` drops the line the room does not need
- rename anything inside it and the slide still finds it
''',
    code('tokenize', trim=['# the deck trims']),
)
```

```python
slide('''
## And what it printed

- `result()` is the cell's own output, not a number typed twice
- re-run the lecture and the slide changes with it
''',
    code('counts', drop=['from collections']),
    result('counts'),
    layout='split',
)
```

```python
slide('''
## And what it drew

- `figure()` stores only the name
- the picture is read out of the lecture when the deck is rendered
- so there is no PNG beside the deck to go stale or be lost
''',
    figure('lengths'),
    layout='split',
)
```

```python
slide('''
## What the build says

- a name that stops matching is **reported**, not guessed at
- so is a bullet that has become a sentence
- `pypresent audit` says whether every slide still fit
''',
    notes='the bit to say out loud, which never reaches the slide',
)
```
