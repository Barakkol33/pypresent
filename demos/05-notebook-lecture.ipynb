```python
# slide: tokenize
STOPWORDS = {'a', 'the', 'of', 'and'}

def tokenize(text):
    # the deck trims this line away
    return [w for w in text.lower().split() if w not in STOPWORDS]

print(tokenize('A short sentence of the corpus'))
```

```python
# slide: counts
from collections import Counter

CORPUS = 'the cat sat on the mat and the cat sat again'
counts = Counter(tokenize(CORPUS))
print(counts.most_common(3))
print(f'{len(counts)} distinct tokens')
```

```python
# slide: lengths
import matplotlib.pyplot as plt

lengths = [len(w) for w in tokenize(CORPUS)]
plt.figure(figsize=(6, 3.2))
plt.hist(lengths, bins=range(1, 8), rwidth=.86, color='#1f5f6b')
plt.title('Token length')
plt.xlabel('characters')
plt.tight_layout()
plt.show()
```
